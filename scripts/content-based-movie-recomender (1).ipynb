```python
movies=pd.read_csv('/kaggle/input/tmdb-movie-metadata/tmdb_5000_movies.csv')
movies.iloc[1,:]
```

```python
credits=pd.read_csv('/kaggle/input/tmdb-movie-metadata/tmdb_5000_credits.csv')
credits.head(5)
```

```python
movies=movies.merge(credits,on='title')
```

```python
movies.dropna(subset=['overview'], inplace=True)
```

```python
import ast

def convert(obj):
    l=[]
    for i in ast.literal_eval(obj):
        l.append(i['name'])
    return l
```

```python
movies['genres']=movies['genres'].apply(convert)
```

```python
movies['keywords']=movies['keywords'].apply(convert)
```

```python
def convert_cast(obj):
    l=[]
    counter=3
    for i in ast.literal_eval(obj):
        if counter<=3 and counter>0:
            l.append(i['name'])
            counter=counter-1
        else:
            break
    return l
```

```python
movies['cast']=movies['cast'].apply(convert_cast)
```

```python
def convert_crew(obj):
    l=[]
    for i in ast.literal_eval(obj):
        if i['job']=='Director':
            l.append(i['name'])
            break
    return l
```

```python
movies['crew']=movies['crew'].apply(convert_crew)
```

```python
movies['overview']=movies['overview'].apply(lambda x:x.split())
```

```python
categorical_colo=['genres','keywords','cast','crew']
movies[categorical_colo]=movies[categorical_colo].applymap(lambda x:[i.replace(" ","") for i in x])
```

```python
movies = movies[['movie_id', 'title', 'overview', 'genres', 'keywords', 'cast', 'crew']]
```

```python
movies.head()
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
new_df = movies[['movie_id', 'title', 'tags']].copy()
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x:" ".join(x))
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x: x.lower())
```

```python
from sentence_transformers import SentenceTransformer

# Load a pre-trained model (all-MiniLM-L6-v2 is fast and accurate)
model = SentenceTransformer('all-MiniLM-L6-v2')

# This takes a few minutes to run on 5000 movies
vectors = model.encode(new_df['tags'].tolist())
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
vectors.shape
```

```python
cosine_similarity(vectors).shape
```

```python
similarity=cosine_similarity(vectors)
```

```python
import pickle
pickle.dump(new_df.to_dict(), open('movie_dict.pkl', 'wb'))
pickle.dump(similarity, open('similarity_matrix.pkl', 'wb'))
```

```python
emotion_to_genre = {
    'happy': ['Animation', 'Adventure', 'Comedy'],
    'sad': ['Drama', 'Romance', 'Musical'],
    'angry': ['Action', 'Thriller', 'Crime'],
    'fear': ['Horror', 'Mystery'],
    'surprise': ['Sci-Fi', 'Fantasy', 'Mystery'],
    'neutral': ['Documentary', 'Biography', 'History'],
    'disgust': ['Family', 'Animation']
}

# 2. Define the Smart Recommender
def recommend_by_mood(movie_title, user_emotion):
    # Find movie index
    idx = new_df[new_df['title'] == movie_title].index[0]

    # Get similarity scores
    distances = similarity[idx]

    # Sort and get top 50 (a larger pool for filtering)
    movie_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:51]

    # Get target genres for the emotion
    target_genres = emotion_to_genre.get(user_emotion, [])

    final_recommendations = []
    for i in movie_list:
        # Fetch genre data from the original 'movies' dataframe
        movie_data = movies.iloc[i[0]]

        # Check if the movie matches the mood genres
        if any(genre in movie_data['genres'] for genre in target_genres):
            final_recommendations.append(movie_data.title)

        # Stop once we have 5 recommendations
        if len(final_recommendations) == 5:
            break

    return final_recommendations

# 3. Test it manually (using a fake emotion for now)
# recommendations = recommend_by_mood('Inception', 'happy')
# print(recommendations)
```
